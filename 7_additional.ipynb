```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# إعداد لوحة الرسم بدقة عالية لتكون واضحة واحترافية
fig, ax = plt.subplots(figsize=(16, 10), dpi=300)
ax.set_facecolor('#f8fafc')
fig.patch.set_facecolor('#ffffff')

# --- 1. العنوان الرئيسي ---
ax.text(0.05, 0.92, "Data Science & MLOps Project", fontsize=12, fontweight='bold', color='#db2777')
ax.text(0.05, 0.85, "E-Commerce Late Delivery Prediction", fontsize=24, fontweight='bold', color='#1e293b')
ax.text(0.05, 0.80, "A modular 6-notebook pipeline to predict delayed orders using the Olist Brazilian E-Commerce dataset.", fontsize=11, color='#64748b')

# --- 2. خطوات تدفق العمل (Pipeline Workflow - 6 Notebooks) ---
steps = [
    ("1. Ingestion", "Read & Join tables\nfrom PostgreSQL"),
    ("2. Labeling", "Create target\n(is_late: 0/1)"),
    ("3. Splitting", "Stratified split\nprevent leakage"),
    ("4. EDA", "Analyze train\nset insights"),
    ("5. Features", "Impute, Scale\n& Save .pkl"),
    ("6. Modeling", "Random Forest\n& Baseline eval")
]

box_width = 0.13
box_height = 0.18
start_x = 0.05
y_pos = 0.55

for i, (title, desc) in enumerate(steps):
    x = start_x + (i * 0.155)
    # رسم الصندوق
    rect = patches.FancyBboxPatch((x, y_pos), box_width, box_height, boxstyle="round,pad=0.02",
                                 facecolor='#ffffff', edgecolor='#cbd5e1', linewidth=1.5)
    ax.add_patch(rect)
    # النصوص داخل الصندوق
    ax.text(x + box_width/2, y_pos + 0.12, title, fontsize=10, fontweight='bold', color='#2563eb', ha='center')
    ax.text(x + box_width/2, y_pos + 0.06, desc, fontsize=8, color='#475569', ha='center', linespacing=1.3)
    
    # رسم سهم بين الصناديق
    if i < len(steps) - 1:
        ax.annotate('', xy=(x + box_width + 0.005, y_pos + box_height/2), 
                    xytext=(x + box_width + 0.02, y_pos + box_height/2),
                    arrowprops=dict(arrowstyle="->", color='#94a3b8', lw=1.5))

# --- 3. قسم معلومات الداتاست والأدوات (Tech Stack & Dataset) ---
# صندوق الداتاست
rect_ds = patches.FancyBboxPatch((0.05, 0.15, 0.43, 0.32), boxstyle="round,pad=0.02", facecolor='#ffffff', edgecolor='#cbd5e1', linewidth=1.5)
ax.add_patch(rect_ds)
ax.text(0.07, 0.42, "📊 Dataset & Architecture Overview", fontsize=12, fontweight='bold', color='#1e293b')
ax.text(0.07, 0.36, "• Dataset: Olist Brazilian E-Commerce (100K+ orders)", fontsize=9, color='#475569')
ax.text(0.07, 0.31, "• Infrastructure: PostgreSQL via Docker container", fontsize=9, color='#475569')
ax.text(0.07, 0.26, "• Methodology: Strict Artifact-based pipeline (Artifacts 1 to 6)", fontsize=9, color='#475569')
ax.text(0.07, 0.21, "• Core Features: total_price, total_payment", fontsize=9, color='#475569')

# صندوق الأدوات المستخدمة (Tech Stack)
rect_tech = patches.FancyBboxPatch((0.52, 0.15, 0.43, 0.32), boxstyle="round,pad=0.02", facecolor='#ffffff', edgecolor='#cbd5e1', linewidth=1.5)
ax.add_patch(rect_tech)
ax.text(0.54, 0.42, "🛠️ Tech Stack & Libraries", fontsize=12, fontweight='bold', color='#1e293b')
ax.text(0.54, 0.36, "• Languages & Core: Python, Pandas, NumPy", fontsize=9, color='#475569')
ax.text(0.54, 0.31, "• Machine Learning: Scikit-Learn (RandomForestClassifier)", fontsize=9, color='#475569')
ax.text(0.54, 0.26, "• Database & Ops: PostgreSQL, Docker, Joblib (.pkl)", fontsize=9, color='#475569')
ax.text(0.54, 0.21, "• Visualization: Matplotlib, Seaborn, Jupyter Notebooks", fontsize=9, color='#475569')

# إخفاء المحاور وضبط الحدود
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')

# حفظ الصورة بدقة عالية
plt.tight_layout()
plt.savefig('my_mlops_project_infographic.png', dpi=300, bbox_inches='tight')
plt.show()

print("تم إنشاء وحفظ صورة الانفوجرافيك الخاصة بمشروعك بنجاح باسم: my_mlops_project_infographic.png")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(14, 16), dpi=300)
ax.set_facecolor('#0f172a')
fig.patch.set_facecolor('#0f172a')

# عناوين ومراحل المخطط
steps = [
    ("PostgreSQL (Docker)", "Source Database: Orders, Items, Payments", "#e17055", "db"),
    ("1_Read_and_Join.ipynb", "Data Ingestion & Aggregation\nArtifact: artifact_1_joined.csv", "#2563eb", "nb"),
    ("2_Create_Labels.ipynb", "Target Definition (is_late: 0/1)\nArtifact: artifact_2_labeled.csv", "#2563eb", "nb"),
    ("3_Data_Split.ipynb", "Stratified Split (Train 80% / Val 10% / Test 10%)\nArtifact: artifact_3_train / val / test.csv", "#2563eb", "nb"),
    ("4_EDA.ipynb", "Exploratory Data Analysis (Train Set Only)\nArtifact: artifact_4_EDA_chart.png", "#7c3aed", "nb"),
    ("5_Feature_Engineering.ipynb", "Imputation, Scaling & Fitted Objects\nArtifacts: .pkl scalers & features.csv", "#2563eb", "nb"),
    ("6_Model_Training.ipynb", "Baseline vs Balanced RandomForest & Evaluation\nArtifact: artifact_6_model.pkl", "#059669", "nb")
]

y_positions = [13, 11, 9, 7, 5, 3, 1]

# رسم العقد والأسهم
for (name, desc, color, n_type), y in zip(steps, y_positions):
    box = patches.FancyBboxPatch((2, y - 0.6), 8, 1.2, boxstyle="round,pad=0.3",
                                 edgecolor=color, facecolor='#1e293b', linewidth=2)
    ax.add_patch(box)
    ax.text(6, y + 0.15, name, color='#f8fafc', fontsize=12, fontweight='bold', ha='center', va='center')
    ax.text(6, y - 0.25, desc, color='#94a3b8', fontsize=9, ha='center', va='center')

# رسم الأسهم التتابعية
for i in range(len(y_positions) - 1):
    y_from = y_positions[i] - 0.6
    y_to = y_positions[i+1] + 0.6
    ax.annotate('', xy=(6, y_to), xytext=(6, y_from),
                arrowprops=dict(arrowstyle="->", color='#38bdf8', lw=2.5, mutation_scale=15))

# ضبط أبعاد الرسم وحفظ الصورة
ax.set_xlim(0, 12)
ax.set_ylim(-0.5, 14.5)
ax.axis('off')
plt.title("MLOps Modular Pipeline Architecture — Brazilian E-Commerce", 
          color='#ffffff', fontsize=15, fontweight='bold', pad=20)

plt.tight_layout()
plt.savefig('pipeline_architecture.png', dpi=300, facecolor=fig.get_facecolor(), edgecolor='none')
plt.show()
print("تم حفظ الصورة بنجاح باسم pipeline_architecture.png بدقة عالية!")
```
