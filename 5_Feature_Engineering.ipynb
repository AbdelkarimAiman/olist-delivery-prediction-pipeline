## Transform

استيراد المكتبات وتحميل الأقسام الثلاثة

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
import joblib

# قراءة أقسام البيانات الثلاثة (Artifact 3)
train_df = pd.read_csv('artifact_3_train.csv')
val_df = pd.read_csv('artifact_3_val.csv')
test_df = pd.read_csv('artifact_3_test.csv')

print("تم تحميل البيانات بنجاح!")
```

استخراج ميزات جديدة من التواريخ

```python
# تحويل أعمدة التواريخ إلى صيغة DateTime
for df in [train_df, val_df, test_df]:
    df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])
    
    # استخراج ميزات جديدة تفيد النموذج
    df['purchase_month'] = df['order_purchase_timestamp'].dt.month
    df['purchase_day_of_week'] = df['order_purchase_timestamp'].dt.dayofweek
    df['purchase_hour'] = df['order_purchase_timestamp'].dt.hour

print("تم استخراج ميزات التواريخ بنجاح!")
```

معالجة القيم الناقصة (Imputation)

```python
# تحديد الأعمدة الرقمية الأساسية
features = ['total_price', 'total_payment']

# تدريب أداة تعويض الفراغات بالوسيط (Median) على التدريب فقط، وتطبيقها على الباقي
imputer = SimpleImputer(strategy='median')
train_df[features] = imputer.fit_transform(train_df[features])
val_df[features] = imputer.transform(val_df[features])
test_df[features] = imputer.transform(test_df[features])

# حفظ أداة الـ Imputer كـ Artifact للإنتاج
joblib.dump(imputer, 'artifact_5_imputer.pkl')
print("تم معالجة القيم الناقصة وحفظ الأداة!")
```

تحجيم البيانات (Scaling) وحفظ ملف البكل (.pkl)

```python
# تدريب أداة التحجيم على التدريب وتطبيقها على التحقق والاختبار
scaler = StandardScaler()
train_df[features] = scaler.fit_transform(train_df[features])
val_df[features] = scaler.transform(val_df[features])
test_df[features] = scaler.transform(test_df[features])

# حفظ أداة التحجيم (ملف الـ .pkl الأساسي للإنتاج)
joblib.dump(scaler, 'artifact_5_scaler.pkl')
print("تم تحجيم البيانات وحفظ ملف الـ Scaler (.pkl) بنجاح!")
```

تحجيم البيانات (Scaling) وحفظ ملف البكل (.pkl)

```python
# تدريب أداة التحجيم على التدريب وتطبيقها على التحقق والاختبار
scaler = StandardScaler()
train_df[features] = scaler.fit_transform(train_df[features])
val_df[features] = scaler.transform(val_df[features])
test_df[features] = scaler.transform(test_df[features])

# حفظ أداة التحجيم (ملف الـ .pkl الأساسي للإنتاج)
joblib.dump(scaler, 'artifact_5_scaler.pkl')
print("تم تحجيم البيانات وحفظ ملف الـ Scaler (.pkl) بنجاح!")
```

حفظ الجداول النهائية وقائمة الميزات

```python
# تحديد قائمة الميزات النهائية التي ستدخل للنموذج
final_features = features + ['purchase_month', 'purchase_day_of_week', 'purchase_hour']

# حفظ الجداول النهائية (Artifact 5)
train_df.to_csv('artifact_5_train_features.csv', index=False)
val_df.to_csv('artifact_5_val_features.csv', index=False)
test_df.to_csv('artifact_5_test_features.csv', index=False)

# حفظ قائمة الميزات في ملف نصي
with open('artifact_5_feature_list.txt', 'w') as f:
    f.write(','.join(final_features))

print("تم حفظ الجداول النهائية وقائمة الميزات بنجاح! النوتبوك الخامس اكتمل بالكامل.")
```
