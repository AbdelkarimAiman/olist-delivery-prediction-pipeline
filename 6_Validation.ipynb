## Train, Tune, Evaluate

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, f1_score
import joblib

# قراءة الجداول الجاهزة
train_df = pd.read_csv('artifact_5_train_features.csv')
val_df = pd.read_csv('artifact_5_val_features.csv')
test_df = pd.read_csv('artifact_5_test_features.csv')

# قراءة قائمة الميزات
with open('artifact_5_feature_list.txt', 'r') as f:
    features = f.read().split(',')

X_train, y_train = train_df[features], train_df['is_late']
X_val, y_val = val_df[features], val_df['is_late']
X_test, y_test = test_df[features], test_df['is_late']

# 1. Baseline بسيط (مثلاً، توقع أن كل الطلبات تصل في الموعد "0")
baseline_preds = [0] * len(y_val)
print("--- تقييم الـ Baseline البسيط على التحقق ---")
print(classification_report(y_val, baseline_preds, zero_division=0))

# 2. تدريب نموذج حقيقي مع مراعاة البيانات غير المتوازنة (class_weight='balanced')
model = RandomForestClassifier(random_state=42, class_weight='balanced')
model.fit(X_train, y_train)

# 3. التقييم على بيانات التحقق (Validation)
val_preds = model.predict(X_val)
print("\n--- تقييم النموذج على التحقق (لضبط الإعدادات) ---")
print(classification_report(y_val, val_preds))

# 4. اللمسة الأخيرة: التقييم على بيانات الاختبار (Test)
test_preds = model.predict(X_test)
print("\n--- التقييم النهائي على الاختبار (Test) ---")
print(classification_report(y_test, test_preds))

# 5. حفظ النموذج النهائي كـ Artifact
joblib.dump(model, 'artifact_6_model.pkl')
print("تم تدريب وحفظ النموذج النهائي بنجاح كـ Artifact 6!")
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, f1_score
import joblib

# Load processed datasets
train_df = pd.read_csv('artifact_5_train_features.csv')
val_df = pd.read_csv('artifact_5_val_features.csv')
test_df = pd.read_csv('artifact_5_test_features.csv')

# Load the feature list
with open('artifact_5_feature_list.txt', 'r') as f:
    features = f.read().split(',')

X_train, y_train = train_df[features], train_df['is_late']
X_val, y_val = val_df[features], val_df['is_late']
X_test, y_test = test_df[features], test_df['is_late']

# 1. Simple Baseline (e.g., predicting that all orders arrive on time "0")
baseline_preds = [0] * len(y_val)
print("--- Simple Baseline Evaluation on Validation Data ---")
print(classification_report(y_val, baseline_preds, zero_division=0))

# 2. Train an actual model handling imbalanced data (class_weight='balanced')
model = RandomForestClassifier(random_state=42, class_weight='balanced')
model.fit(X_train, y_train)

# 3. Evaluation on Validation Data
val_preds = model.predict(X_val)
print("\n--- Model Evaluation on Validation (Hyperparameter Tuning) ---")
print(classification_report(y_val, val_preds))

# 4. Final Touch: Evaluation on Test Data
test_preds = model.predict(X_test)
print("\n--- Final Evaluation on Test Data ---")
print(classification_report(y_test, test_preds))

# 5. Save the final model as an Artifact
joblib.dump(model, 'artifact_6_model.pkl')
print("Final model trained and successfully saved as Artifact 6!")
```
