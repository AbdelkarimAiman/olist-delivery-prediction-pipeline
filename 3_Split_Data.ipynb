## Train / Validation / Test Split

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv('artifact_2_labeled_data.csv')

# 1. تقسيم البيانات (80% تدريب، 20% مبدئي للاختبار والتحقق)
# نستخدم stratify للحفاظ على توازن فئة "التأخير" في كل الأقسام
train_df, temp_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df['is_late'])

# 2. تقسيم الجزء المتبقي إلى نصفين (10% تحقق، 10% اختبار)
val_df, test_df = train_test_split(temp_df, test_size=0.5, random_state=42, stratify=temp_df['is_late'])

# 3. حفظ الـ Artifacts
train_df.to_csv('artifact_3_train.csv', index=False)
val_df.to_csv('artifact_3_val.csv', index=False)
test_df.to_csv('artifact_3_test.csv', index=False)
```
