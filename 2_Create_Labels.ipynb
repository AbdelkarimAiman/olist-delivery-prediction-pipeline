## Create Labels

```python
import pandas as pd
import numpy as np

# قراءة البيانات المدمجة من الخطوة السابقة
df = pd.read_csv('artifact_1_joined_data.csv')

# تحويل نصوص التواريخ إلى صيغة تاريخ فعلية
df['delivered_date'] = pd.to_datetime(df['order_delivered_customer_date'])
df['estimated_date'] = pd.to_datetime(df['order_estimated_delivery_date'])

# إنشاء الهدف: 1 يعني متأخر، 0 يعني في الموعد
df['is_late'] = np.where(df['delivered_date'] > df['estimated_date'], 1, 0)

# حفظ الملف للخطوة القادمة
df.to_csv('artifact_2_labeled_data.csv', index=False)
print("تم إنشاء عمود الهدف وحفظ Artifact 2 بنجاح!")
```
